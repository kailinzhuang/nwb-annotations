## Stimulus data

```python
# list of stimuli names
stims_keys = list(nwb.stimulus.keys())
print(stims_keys)
```

```python
# let's look at one of them
stim_idx = 20
stim_name = stims_keys[stim_idx]
print(f"Name of stimulus: {stim_name}")

# get the stimulus data from nwb file
stim_data = np.asarray(nwb.stimulus[stim_name].data)

# plot spectrogram:
# first we set up the parameters needed for processing audio and making spectrogramns
audio_rate = nwb.acquisition['audio'].rate
stim_fs = audio_rate

# spectrogram sample rate
stim_sample_rate = 1000

# make dict to make spectrogram
stim_params=dict()
stim_params['fband'] = 120
stim_params['nstd'] = 6
stim_params['high_freq'] = 8000
stim_params['low_freq'] = 250
stim_params['log'] = 1
stim_params['stim_rate'] = stim_sample_rate
tfrep = timefreq_raw(stim_data, stim_fs, 'ft', stim_params)
stim = {
    'type': 'tfrep',
    'rawFile': stim_name,
    'tfrep': tfrep,
    'rawSampleRate': tfrep['params']['rawSampleRate'],
    'sampleRate': stim_sample_rate,
    'stimLength': tfrep['spec'].shape[1] / stim_sample_rate,
    'nStimChannels' : tfrep['f'].shape[0],
    'maxStimAmp': np.max(tfrep["spec"])
}

# plot spectrogram
DBNOISE = 80
minSpect = tfrep["spec"].max() - DBNOISE
maxB = tfrep["spec"].max()

f, (ax1, ax2) = plt.subplots(2, 1, sharex=True, dpi=100, figsize = (8,4))


minSpect = tfrep['spec'].max()-DBNOISE
maxB = tfrep['spec'].max()
im = ax1.imshow(tfrep['spec'], extent=[tfrep['t'][0], tfrep['t'][-1], tfrep['f'][0]*1e-3, tfrep['f'][-1]*1e-3],
                aspect='auto', interpolation='nearest', origin='lower', cmap=cmap, vmin=minSpect, vmax=maxB)
ax1.set_ylim(0, 8)
ax1.set_ylabel('Frequency (kHz)')

tval = np.arange(stop=len(stim_data))/audio_rate
ax2.plot(tval, stim_data.flatten(), 'k', alpha=0.7)
ax2.set_xlabel('Time (s)')
# plt.colorbar(im)
```

## Maybe one thing we can do is to use the start of first event as time 0
Then we can use 'relative time' to make indexing easier later on.


```python
# add another column to ann for relative timing to first start
ann["relative_t_start"] = ann["project.t_start"] - starts[0]
ann["relative_t_stop"] = ann["project.t_stop"] - starts[0]

# same for indices
ann["relative_start_index"] = ann["project.start_index"] - starts_indices[0]
ann["relative_stop_index"] = ann["project.stop_index"] - starts_indices[0]

ann
```

# Annotations spectrogram

1. we can generate a spectrogram for the duration of annotation (from first ann to last ann)
2. for each annotation, we can index the spectrogram

```python
# find mic_data from first ann to last ann
start_index = starts_indices[0]
stop_index = stops_indices[-1]

mic_data_segment = mic_data[start_index:stop_index]
mic_data_segment.shape
```

```python
# how long is this?
hms = str(datetime.timedelta(seconds=mic_data_segment.shape[0]/audio_rate))
print(f"length of mic_data_segment: {hms}")
```

```python
ann_tfrep = timefreq_raw(mic_data_segment, audio_rate, 'ft', stim_params)
```

```python
ann_spec = ann_tfrep['spec']
print(f"shape of ann_spec: {ann_spec.shape}")
```

The shape of ann_spec is freq x time.
The time axis should correspond to the time of mic_data_segment, but they will be in different units.
We can look at what this change of unit is:

```python
print(f"spectrogram dt:  {ann_tfrep['t'][2]-ann_tfrep['t'][1]} s")
```

```python
# now we add red and blue arrows on the spectrogram to show where the annotations are (onset and offset)

# what range of spectrogram to plot?
plot_s_start = 13
plot_s_stop = 17

plot_idx_start = int(plot_s_start * stim_sample_rate)
plot_idx_stop = int(plot_s_stop * stim_sample_rate)

# now find all the events that are in this range
# we cn use relative_t_start and relative_t_stop since the time is relative to the first annotation start
plot_events = ann[(ann["relative_t_start"] >= plot_s_start) & (ann["relative_t_stop"] <= plot_s_stop)]
print(f"number of events in this range: {plot_events.shape[0]}")

fig, ax = plt.subplots(1, 1, dpi=100, figsize = (8,2))
ax.imshow(ann_spec[:, plot_idx_start:plot_idx_stop], 
          extent=[ann_tfrep['t'][plot_idx_start], ann_tfrep['t'][plot_idx_stop], ann_tfrep['f'][0]*1e-3, ann_tfrep['f'][-1]*1e-3],
          aspect='auto', interpolation='nearest', origin='lower', cmap=cmap, vmin=ann_spec.max()-DBNOISE, vmax=ann_spec.max()
        )
ax.set_ylim(0, 8)
ax.set_ylabel('Frequency (kHz)')  
ax.set_xlabel('Time (s)')

for idx, row in plot_events.iterrows():
    ax.arrow(row["relative_t_start"], 8, 0, -1.5, color='r', width=0.001, head_width=0.025, head_length=0.4)
    ax.arrow(row["relative_t_stop"], 8, 0, -1.5, color='b', width=0.001, head_width=0.025, head_length=0.4)
```

```python
# for each annotation, index the spectrogram
plot_n_start = 30
plot_n_stop = 50
plot_n = plot_n_stop - plot_n_start

fig, axes = plt.subplots(plot_n//5, 5, dpi=100, figsize = (6, 2*plot_n//5), sharey=True)

ann_plot = ann.iloc[plot_n_start:plot_n_stop]
for i, (idx, row) in enumerate(ann_plot.iterrows()):
    ann_start_time = row["relative_t_start"]
    ann_stop_time = row["relative_t_stop"]
    
    
    # find the corresponding spectrogram indices
    spec_start_idx = int(ann_start_time * stim_sample_rate)
    spec_stop_idx = int(ann_stop_time * stim_sample_rate)
    
    # plt
    pad = 5    # pad by this many samples on either side
    plot_idx_start = max(0, spec_start_idx - pad)
    plot_idx_stop = min(ann_spec.shape[1], spec_stop_idx + pad)
    
    ax = axes[i//5, i%5]
    ax.imshow(ann_spec[:, plot_idx_start:plot_idx_stop], 
              extent=[ann_tfrep['t'][plot_idx_start], ann_tfrep['t'][plot_idx_stop], ann_tfrep['f'][0]*1e-3, ann_tfrep['f'][-1]*1e-3],
              aspect='auto', interpolation='nearest', origin='lower', cmap=cmap, vmin=ann_spec.max()-DBNOISE, vmax=ann_spec.max()
            )

    ax.set_xticks([ann_start_time, ann_stop_time])
    ax.set_xticklabels([f'{ann_start_time:.2f}', f'{ann_stop_time:.2f}'])

fig.supxlabel('Time (s) (from start of first annotation)')
fig.supylabel('Frequency (kHz)')
fig.tight_layout()
```
